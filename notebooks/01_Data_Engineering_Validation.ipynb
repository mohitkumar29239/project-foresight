### 10.1.4 Identifier Validation

Identifier columns uniquely describe each product-store combination and play a critical role in data integration and feature engineering. This subsection validates the identifier columns by checking for missing values, uniqueness, and cardinality.

> **📌 Objective:** Verify that the identifier columns are complete and suitable for downstream data integration.

### ⚙️ 11.1.3 Configure the Development Sales Scope

The complete M5 sales dataset contains more than 30,000 product-store time series and would generate over 58 million records after conversion from wide to long format. Processing the complete dataset may require substantial memory, storage, and execution time on a local system.

To support efficient and reproducible local development, a category-balanced subset of 30 products—10 from each category—was selected using deterministic sorting, while retaining all 10 stores and all 1,913 historical sales days.

This subset supports pipeline development across product categories, stores, states and time periods, but it should not be interpreted as a statistically representative sample of the complete product or department population.

**Objective:**

- Define a reproducible and laptop-friendly sales development scope.
- Select an equal and reproducible number of development products from each product category.
- Retain all available stores and historical sales days.

````python
# ============================================================
# 11.1.3 Configure the Development Sales Scope
# ============================================================
# Define the number of products to select from each product category.
PRODUCTS_PER_CATEGORY = 10

# Display the configuration settings for the development dataset.
print("Development Scope Configuration")
print("-" * 40)

print(f"Products selected per category : {PRODUCTS_PER_CATEGORY}")
print(f"Available product categories   : {sales_daily['cat_id'].nunique()}")
print(f"Available retail stores        : {sales_daily['store_id'].nunique()}")
print(f"Historical sales-day columns   : {len(sales_day_columns)}")
````

### 🔄 11.1.4 Reshape the Sales Dataset to Long Format

The development sales dataset currently stores each historical sales day in a separate column. Although this wide structure is suitable for compact storage, it is not convenient for time-series analysis, relational joins, feature engineering, or downstream analytical workflows.

The dataset is therefore converted into long format, where each row represents the units sold for one SKU-store series on one historical sales day. This structure will support subsequent integration with calendar and pricing data.

**Objective:**

- Convert the wide sales dataset into a row-level daily sales structure.
- Preserve all product, store, category, department, and state identifiers.
- Create standardized `day` and `units_sold` fields.
- Validate the expected number and uniqueness of daily sales records.

````python
# ============================================================
# Reshape the Development Sales Dataset to Long Format
# ============================================================

# Convert the historical sales-day columns into row-level records
# while preserving all product and store identifiers.
sales_long = sales_daily_subset.melt(
    id_vars=identifier_columns,
    value_vars=sales_day_columns,
    var_name="day",
    value_name="units_sold"
)

# Confirm that the long-format dataset was created.
print(f"Long-format dataset created: {sales_long.shape}")
````

### 🛠️ 11.1.8 Engineer Revenue and Promotion Fields

Revenue and promotional indicators provide valuable business context beyond sales volume. This step derives additional features required for sales analysis and forecasting.

This step will:

- calculate revenue for each sales record;
- create a promotion indicator when the current weekly selling price is lower than the previous available weekly price;
- preserve all existing sales information;
- validate the engineered business features.

The resulting dataset will contain the core measures required for the final sales dataset.

````python
# ============================================================
# Engineer Revenue and Promotion Fields
# ============================================================

sales_features = sales_priced.copy()

# Calculate daily revenue.
sales_features["revenue"] = (
    sales_features["units_sold"]
    * sales_features["sell_price"]
)

# Create one price record per SKU and retail week.
weekly_prices = (
    sales_features[
        ["sku_id", "wm_yr_wk", "sell_price"]
    ]
    .drop_duplicates()
    .sort_values(["sku_id", "wm_yr_wk"])
    .reset_index(drop=True)
)

# Treat zero prices as unavailable when comparing prices.
weekly_prices["valid_price"] = (
    weekly_prices["sell_price"]
    .where(weekly_prices["sell_price"] > 0)
)

# Obtain the previous available weekly price.
weekly_prices["previous_price"] = (
    weekly_prices
    .groupby("sku_id")["valid_price"]
    .ffill()
    .groupby(weekly_prices["sku_id"])
    .shift(1)
)

# Mark a week as promotional when its valid price decreases.
weekly_prices["promo_flag"] = (
    weekly_prices["valid_price"].notna()
    & weekly_prices["previous_price"].notna()
    & (
        weekly_prices["valid_price"]
        < weekly_prices["previous_price"]
    )
).astype("int8")

# Attach the weekly promotion flag to every daily record.
sales_features = sales_features.drop(
    columns="promo_flag",
    errors="ignore"
).merge(
    weekly_prices[
        ["sku_id", "wm_yr_wk", "promo_flag"]
    ],
    on=["sku_id", "wm_yr_wk"],
    how="left",
    validate="many_to_one"
)

print("Revenue and promotion features engineered successfully.")
print(f"Dataset shape: {sales_features.shape}")
print(
    "Promotional daily records: "
    f"{sales_features['promo_flag'].sum():,}"
)
print(
    "Promotional SKU-weeks: "
    f"{weekly_prices['promo_flag'].sum():,}"
)
````

````python
# ============================================================
# Validate Revenue and Promotion Features
# ============================================================

expected_rows = len(sales_priced)
actual_rows = len(sales_features)

missing_revenue = sales_features["revenue"].isna().sum()
missing_promo = sales_features["promo_flag"].isna().sum()

negative_revenue = (sales_features["revenue"] < 0).sum()

invalid_promo = (
    ~sales_features["promo_flag"].isin([0, 1])
).sum()

assert actual_rows == expected_rows
assert missing_revenue == 0
assert missing_promo == 0
assert negative_revenue == 0
assert invalid_promo == 0

print("✅ Revenue and promotion features validated successfully.")
print(f"Missing revenue values : {missing_revenue:,}")
print(f"Missing promo flags    : {missing_promo:,}")
print(f"Negative revenue rows  : {negative_revenue:,}")
print(f"Promotional daily rows : {sales_features['promo_flag'].sum():,}")
````

Validation confirmed that:

#### 📅 11.3.4 Engineer SKU Lifecycle Fields

SKU lifecycle attributes are derived from historical price availability and positive sales activity. These fields describe when each item-store combination became commercially observable, when it first and last recorded sales, and how long it existed within the observed sales period.

Because the source data does not contain an explicit product-launch date, `launch_date` is treated as an operational proxy defined as the first calendar date of the earliest retail week containing a positive selling price.

This step will:

- prepare a complete mapping between historical sales-day columns and calendar dates;
- estimate the launch date for each SKU from its earliest positive-price week;
- identify the first and last dates on which each SKU recorded positive sales;
- calculate SKU age at the end of the observed sales period;
- validate lifecycle completeness, chronological order, identifier integrity, and record preservation.

The validated lifecycle attributes will support SKU-status classification, inventory planning, and downstream business analysis.

````python
# ============================================================
# Prepare SKU Lifecycle References
# ============================================================

# Remove previously engineered lifecycle fields when the
# subsection is rerun during notebook development.
existing_lifecycle_columns = [
    "launch_date",
    "first_sale_date",
    "last_sale_date",
    "product_age_days",
    "sku_age_days"
]

sku_master = sku_master.drop(
    columns=[
        column
        for column in existing_lifecycle_columns
        if column in sku_master.columns
    ],
    errors="ignore"
)

# Preserve the foundation row count for merge validation.
lifecycle_input_rows = len(sku_master)

# Identify and chronologically order the historical sales-day columns.
day_columns = sorted(
    [
        column
        for column in sku_sales.columns
        if column.startswith("d_")
    ],
    key=lambda column: int(
        column.split("_")[1]
    )
)

# Confirm that historical sales-day columns were identified.
assert len(day_columns) > 0, (
    "No historical sales-day columns were found."
)

# Extract the numeric day identifiers.
day_numbers = [
    int(column.split("_")[1])
    for column in day_columns
]

# Define the expected uninterrupted historical day sequence.
expected_day_numbers = list(
    range(1, max(day_numbers) + 1)
)

# Confirm that no historical sales-day columns are missing.
assert day_numbers == expected_day_numbers, (
    "The historical sales-day sequence is incomplete."
)

# Create an ordered mapping between sales-day columns and dates.
sales_date_lookup = (
    sku_calendar
    .set_index("d")["date"]
    .reindex(day_columns)
)

# Count sales-day identifiers without a matching Calendar date.
missing_sales_date_mappings = (
    sales_date_lookup
    .isna()
    .sum()
)

# Confirm that the mapped dates follow chronological order.
sales_dates_are_chronological = (
    sales_date_lookup
    .is_monotonic_increasing
)

# Validate the historical day-to-date mapping.
assert missing_sales_date_mappings == 0, (
    f"Missing sales-day date mappings detected: "
    f"{missing_sales_date_mappings}."
)

assert sales_dates_are_chronological, (
    "The mapped historical sales dates are not chronological."
)

# Define the observed historical sales period.
observation_start_date = sales_date_lookup.min()
observation_end_date = sales_date_lookup.max()

print("✅ SKU lifecycle references prepared successfully.")
print(f"Daily sales columns        : {len(day_columns):,}")
print(f"Missing date mappings      : {missing_sales_date_mappings:,}")
print(f"First observed sales date  : {observation_start_date.date()}")
print(f"Last observed sales date   : {observation_end_date.date()}")
````

````python
# ============================================================
# Engineer SKU Launch Date
# ============================================================

# Identify the earliest retail week containing
# a positive selling price for each item-store combination.
sku_launch_weeks = (
    sku_prices.loc[
        sku_prices["sell_price"] > 0,
        [
            "store_id",
            "item_id",
            "wm_yr_wk"
        ]
    ]
    .groupby(
        ["store_id", "item_id"],
        as_index=False
    )["wm_yr_wk"]
    .min()
    .rename(
        columns={
            "wm_yr_wk": "launch_week"
        }
    )
)

# Determine the first available Calendar date
# associated with every retail week.
week_start_dates = (
    sku_calendar
    .groupby(
        "wm_yr_wk",
        as_index=False
    )["date"]
    .min()
    .rename(
        columns={
            "wm_yr_wk": "launch_week",
            "date": "launch_date"
        }
    )
)

# Convert each earliest priced week into
# an estimated SKU launch date.
sku_launch_dates = sku_launch_weeks.merge(
    week_start_dates,
    on="launch_week",
    how="left",
    validate="many_to_one"
)

# Count launch weeks without a Calendar-date mapping.
missing_launch_date_mappings = (
    sku_launch_dates["launch_date"]
    .isna()
    .sum()
)

# Confirm that each item-store combination
# has only one launch-date record.
duplicate_launch_keys = (
    sku_launch_dates[
        ["store_id", "item_id"]
    ]
    .duplicated()
    .sum()
)

assert missing_launch_date_mappings == 0, (
    f"Missing launch-date mappings detected: "
    f"{missing_launch_date_mappings}."
)

assert duplicate_launch_keys == 0, (
    f"Duplicate SKU launch records detected: "
    f"{duplicate_launch_keys}."
)

# Attach the estimated launch date
# to the SKU Master foundation.
sku_master = sku_master.merge(
    sku_launch_dates[
        [
            "store_id",
            "item_id",
            "launch_date"
        ]
    ],
    on=["store_id", "item_id"],
    how="left",
    validate="one_to_one"
)

# Confirm that the merge preserved all foundation records.
assert len(sku_master) == lifecycle_input_rows, (
    "The launch-date merge changed the SKU Master row count."
)

# Confirm that every SKU received a launch date.
missing_launch_dates = (
    sku_master["launch_date"]
    .isna()
    .sum()
)

assert missing_launch_dates == 0, (
    f"SKUs without launch dates detected: "
    f"{missing_launch_dates}."
)

print("✅ SKU launch dates engineered successfully.")
print(f"Earliest launch date       : {sku_master['launch_date'].min().date()}")
print(f"Latest launch date         : {sku_master['launch_date'].max().date()}")
print(f"Missing launch dates       : {missing_launch_dates:,}")
````

````python
# ============================================================
# Engineer First and Last Sale Dates
# ============================================================

# Convert the ordered calendar dates into a NumPy date array.
sales_dates = sales_date_lookup.to_numpy(
    dtype="datetime64[ns]"
)

# Create canonical SKU identifiers using
# each item-store combination.
sales_lifecycle_sku_ids = (
    sku_sales["item_id"].astype("string")
    + "_"
    + sku_sales["store_id"].astype("string")
)

# Confirm that every Sales record represents one unique SKU.
assert sales_lifecycle_sku_ids.is_unique, (
    "Duplicate canonical SKU identifiers were found "
    "in the Sales working dataset."
)

# Define the number of SKU sales-series records.
record_count = len(sku_sales)

# Create empty arrays for first and last sale dates.
first_sale_dates = np.full(
    record_count,
    np.datetime64("NaT"),
    dtype="datetime64[ns]"
)

last_sale_dates = np.full(
    record_count,
    np.datetime64("NaT"),
    dtype="datetime64[ns]"
)

# Process the wide sales dataset in manageable chunks.
chunk_size = 5_000

for start_index in range(
    0,
    record_count,
    chunk_size
):
    end_index = min(
        start_index + chunk_size,
        record_count
    )

    # Extract historical daily sales for the current chunk.
    sales_chunk = sku_sales.iloc[
        start_index:end_index
    ][day_columns].to_numpy()

    # Identify dates containing positive sales.
    positive_sales = sales_chunk > 0

    # Identify SKUs containing at least one positive sale.
    has_sales = positive_sales.any(axis=1)

    # Locate the first positive-sale position.
    first_positions = positive_sales.argmax(axis=1)

    # Locate the last positive-sale position.
    last_positions = (
        positive_sales.shape[1]
        - 1
        - positive_sales[:, ::-1].argmax(axis=1)
    )

    # Create the corresponding dataset row positions.
    chunk_rows = np.arange(
        start_index,
        end_index
    )

    valid_rows = chunk_rows[has_sales]

    # Assign the first positive-sale dates.
    first_sale_dates[valid_rows] = sales_dates[
        first_positions[has_sales]
    ]

    # Assign the last positive-sale dates.
    last_sale_dates[valid_rows] = sales_dates[
        last_positions[has_sales]
    ]

# Create the lifecycle-date table using
# the canonical Project FORESIGHT SKU identifier.
sku_sale_dates = pd.DataFrame(
    {
        "sku_id": sales_lifecycle_sku_ids.to_numpy(),
        "first_sale_date": pd.to_datetime(
            first_sale_dates
        ),
        "last_sale_date": pd.to_datetime(
            last_sale_dates
        )
    }
)

# Confirm that the lifecycle table contains
# one record for every canonical SKU.
assert sku_sale_dates["sku_id"].is_unique, (
    "Duplicate SKU identifiers were found "
    "in the lifecycle-date table."
)

# Remove lifecycle columns created during
# an earlier notebook execution.
sku_master = sku_master.drop(
    columns=[
        "first_sale_date",
        "last_sale_date"
    ],
    errors="ignore"
)

# Preserve the current row count before merging.
lifecycle_rows_before_merge = len(sku_master)

# Attach first and last sale dates
# using the canonical SKU identifier.
sku_master = sku_master.merge(
    sku_sale_dates,
    on="sku_id",
    how="left",
    validate="one_to_one"
)

# Confirm that the merge preserved every SKU record.
assert len(sku_master) == lifecycle_rows_before_merge, (
    "The lifecycle-date merge changed "
    "the SKU Master row count."
)

# Count missing lifecycle dates after the merge.
missing_first_sale_dates = (
    sku_master["first_sale_date"]
    .isna()
    .sum()
)

missing_last_sale_dates = (
    sku_master["last_sale_date"]
    .isna()
    .sum()
)

# Confirm that every SKU received lifecycle dates.
assert missing_first_sale_dates == 0, (
    f"Missing first-sale dates detected: "
    f"{missing_first_sale_dates}."
)

assert missing_last_sale_dates == 0, (
    f"Missing last-sale dates detected: "
    f"{missing_last_sale_dates}."
)

print("✅ First and last sale dates engineered successfully.")
print(f"Lifecycle records created : {len(sku_sale_dates):,}")
print(f"Missing first-sale dates  : {missing_first_sale_dates:,}")
print(f"Missing last-sale dates   : {missing_last_sale_dates:,}")
print(f"Rows preserved            : {len(sku_master):,}")
````

#### 🛠️ 11.3.5 Engineer Pricing Fields

SKU-level pricing attributes are derived from the historical weekly selling prices available for each item-store combination.

The `avg_price` field represents the unweighted average of the available weekly selling-price records, while `max_observed_price` represents the highest selling price recorded during the available pricing history. It should be interpreted as a historical reference-price proxy rather than a confirmed official list price.

This step will:

- validate the completeness and positivity of the source selling-price records;
- calculate the average historical weekly selling price for each SKU;
- determine the maximum observed selling price for each SKU;
- create canonical SKU identifiers for the aggregated pricing records;
- integrate the pricing attributes into the SKU Master dataset;
- validate pricing coverage, logical consistency, and merge integrity.

The validated pricing attributes will support profitability analysis, cost estimation, and downstream business reporting.

````python
# ============================================================
# Prepare and Aggregate SKU Pricing Records
# ============================================================

# Remove previously engineered pricing fields when this
# subsection is rerun during notebook development.
existing_pricing_columns = [
    "avg_price",
    "list_price",
    "max_observed_price"
]

sku_master = sku_master.drop(
    columns=[
        column
        for column in existing_pricing_columns
        if column in sku_master.columns
    ],
    errors="ignore"
)

# Preserve the current SKU Master row count
# for subsequent merge-integrity validation.
pricing_input_rows = len(sku_master)

# Count missing weekly selling-price values.
missing_source_prices = (
    sku_prices["sell_price"]
    .isna()
    .sum()
)

# Count zero or negative weekly selling-price records.
non_positive_source_prices = (
    sku_prices["sell_price"] <= 0
).sum()

# Confirm that the source pricing records are complete.
assert missing_source_prices == 0, (
    f"Missing source selling prices detected: "
    f"{missing_source_prices}."
)

# Confirm that all source selling prices are positive.
assert non_positive_source_prices == 0, (
    f"Non-positive source selling prices detected: "
    f"{non_positive_source_prices}."
)

# Calculate historical pricing metrics
# for every item-store combination.
pricing_metrics = (
    sku_prices
    .groupby(
        ["store_id", "item_id"],
        as_index=False
    )
    .agg(
        avg_price=(
            "sell_price",
            "mean"
        ),
        max_observed_price=(
            "sell_price",
            "max"
        ),
        price_week_count=(
            "sell_price",
            "size"
        )
    )
)

# Round the monetary pricing metrics
# to two decimal places.
pricing_metrics["avg_price"] = (
    pricing_metrics["avg_price"]
    .round(2)
)

pricing_metrics["max_observed_price"] = (
    pricing_metrics["max_observed_price"]
    .round(2)
)

# Create the canonical Project FORESIGHT SKU identifier
# for every aggregated pricing record.
pricing_metrics["sku_id"] = (
    pricing_metrics["item_id"].astype("string")
    + "_"
    + pricing_metrics["store_id"].astype("string")
)

# Count duplicate canonical identifiers
# in the aggregated pricing table.
duplicate_pricing_sku_ids = (
    pricing_metrics["sku_id"]
    .duplicated()
    .sum()
)

# Confirm that one pricing record was created per SKU.
assert duplicate_pricing_sku_ids == 0, (
    f"Duplicate aggregated pricing SKU identifiers detected: "
    f"{duplicate_pricing_sku_ids}."
)

print("✅ SKU pricing metrics calculated successfully.")
print(f"Pricing SKU records       : {len(pricing_metrics):,}")
print(f"Missing source prices     : {missing_source_prices:,}")
print(f"Non-positive source prices: {non_positive_source_prices:,}")
print(f"Duplicate pricing SKU IDs : {duplicate_pricing_sku_ids:,}")
````

````python
# ============================================================
# Integrate Pricing Attributes into the SKU Master
# ============================================================

# Select the pricing attributes required
# for integration with the SKU Master.
sku_pricing_attributes = pricing_metrics[
    [
        "sku_id",
        "avg_price",
        "max_observed_price"
    ]
].copy()

# Merge the pricing attributes using the canonical
# Project FORESIGHT SKU identifier.
sku_master = sku_master.merge(
    sku_pricing_attributes,
    on="sku_id",
    how="left",
    validate="one_to_one"
)

# Confirm that the merge preserved the SKU Master population.
pricing_rows_preserved = (
    len(sku_master)
    == pricing_input_rows
)

# Count SKU Master records without matched pricing attributes.
unmatched_master_pricing = (
    sku_master[
        [
            "avg_price",
            "max_observed_price"
        ]
    ]
    .isna()
    .any(axis=1)
    .sum()
)

# Count aggregated pricing SKUs not represented
# in the SKU Master foundation.
unused_pricing_skus = (
    ~pricing_metrics["sku_id"].isin(
        sku_master["sku_id"]
    )
).sum()

# Prevent continuation when the merge changes
# the SKU Master row count.
assert pricing_rows_preserved, (
    "The pricing merge changed the SKU Master row count."
)

# Confirm that every SKU received its pricing attributes.
assert unmatched_master_pricing == 0, (
    f"SKU Master records without pricing attributes detected: "
    f"{unmatched_master_pricing}."
)

# Confirm that every aggregated pricing SKU
# is represented in the SKU Master.
assert unused_pricing_skus == 0, (
    f"Pricing SKUs not found in the SKU Master detected: "
    f"{unused_pricing_skus}."
)

print("✅ Pricing attributes integrated successfully.")
print(f"SKU Master shape          : {sku_master.shape}")
print(f"SKUs updated              : {len(sku_master):,}")
print(f"Unmatched Master SKUs     : {unmatched_master_pricing:,}")
print(f"Unused pricing SKUs       : {unused_pricing_skus:,}")
print(f"Row count preserved       : {pricing_rows_preserved}")
````

#### 💰 11.3.6 Estimate Cost and Margin Fields

Estimated cost and margin attributes are derived from the engineered `max_observed_price` using a documented business assumption.

Because actual procurement-cost data is unavailable, the maximum observed selling price is used as a reference-price proxy. The estimated unit cost is assumed to represent 70% of this reference price, while the remaining amount represents the estimated gross margin per unit.

These fields are analytical estimates and should not be interpreted as actual procurement costs or realized accounting margins.

This step will:

- define and validate the cost-estimation assumption;
- estimate the unit cost for every canonical SKU;
- calculate the corresponding estimated gross margin per unit;
- validate the calculations, financial relationships, and record preservation.

The validated estimated financial attributes will support profitability-oriented analysis, inventory valuation assumptions, and downstream business reporting.

````python
# ============================================================
# Estimate SKU Cost and Margin Fields
# ============================================================

# Remove previously engineered financial fields when this
# subsection is rerun during notebook development.
existing_financial_columns = [
    "unit_cost",
    "gross_margin",
    "estimated_unit_cost",
    "estimated_gross_margin"
]

sku_master = sku_master.drop(
    columns=[
        column
        for column in existing_financial_columns
        if column in sku_master.columns
    ],
    errors="ignore"
)

# Preserve the current SKU Master row count
# for subsequent integrity validation.
financial_input_rows = len(sku_master)

# Define the documented cost-estimation assumption.
ESTIMATED_COST_RATE = 0.70

# Confirm that the assumption represents
# a valid proportion between zero and one.
assert 0 < ESTIMATED_COST_RATE < 1, (
    "The estimated cost rate must be "
    "greater than 0 and less than 1."
)

# Confirm that the required reference-price field is available.
assert "max_observed_price" in sku_master.columns, (
    "The max_observed_price field is required "
    "before estimating cost and margin."
)

# Count missing reference-price values.
missing_reference_prices = (
    sku_master["max_observed_price"]
    .isna()
    .sum()
)

# Count zero or negative reference-price values.
non_positive_reference_prices = (
    sku_master["max_observed_price"] <= 0
).sum()

# Validate the reference-price field
# before calculating financial estimates.
assert missing_reference_prices == 0, (
    f"Missing reference prices detected: "
    f"{missing_reference_prices}."
)

assert non_positive_reference_prices == 0, (
    f"Non-positive reference prices detected: "
    f"{non_positive_reference_prices}."
)

# Estimate unit cost using the documented
# percentage of the maximum observed price.
sku_master["estimated_unit_cost"] = (
    sku_master["max_observed_price"]
    * ESTIMATED_COST_RATE
).round(2)

# Calculate the estimated gross margin per unit
# using the reference price and estimated unit cost.
sku_master["estimated_gross_margin"] = (
    sku_master["max_observed_price"]
    - sku_master["estimated_unit_cost"]
).round(2)

# Display the financial-engineering summary.
print("✅ Estimated cost and margin fields engineered successfully.")
print(f"Cost estimation rate     : {ESTIMATED_COST_RATE:.0%}")
print(f"Dataset shape            : {sku_master.shape}")
print(f"SKUs updated             : {len(sku_master):,}")
print(f"Missing reference prices : {missing_reference_prices:,}")
````

#### 🏷️ 11.3.7 Classify SKU Status

SKU status is classified using the recency of positive sales activity for each item-store combination.

The classification represents a historical snapshot as of the final date in the observed sales period. It does not represent the SKU’s current real-world commercial status.

The following Project FORESIGHT business rules are applied:

- **Active:** the last recorded sale occurred within the previous 30 days;
- **Slow Moving:** the last recorded sale occurred between 31 and 90 days ago;
- **Inactive:** the last recorded sale occurred more than 90 days ago.

This step will:

- calculate the number of days since the last recorded sale for each SKU;
- classify each SKU using documented recency thresholds;
- retain the recency measure for reporting and classification traceability;
- validate classification completeness, rule accuracy, identifier uniqueness, and record preservation.

The validated SKU-status attributes will support inventory monitoring, assortment analysis, and downstream business reporting.

````python
# ============================================================
# Classify SKU Status
# ============================================================

# Remove previously engineered status fields when this
# subsection is rerun during notebook development.
existing_status_columns = [
    "days_since_last_sale",
    "product_status",
    "sku_status"
]

sku_master = sku_master.drop(
    columns=[
        column
        for column in existing_status_columns
        if column in sku_master.columns
    ],
    errors="ignore"
)

# Preserve the current SKU Master row count.
status_input_rows = len(sku_master)

# Define the classification thresholds.
ACTIVE_MAX_DAYS = 30
SLOW_MOVING_MAX_DAYS = 90

# Validate the threshold sequence.
assert 0 <= ACTIVE_MAX_DAYS < SLOW_MOVING_MAX_DAYS, (
    "The SKU-status thresholds are not logically ordered."
)

# Confirm that the required lifecycle fields are available.
required_status_columns = [
    "sku_id",
    "last_sale_date"
]

missing_status_columns = [
    column
    for column in required_status_columns
    if column not in sku_master.columns
]

assert len(missing_status_columns) == 0, (
    f"Required status fields are missing: "
    f"{missing_status_columns}."
)

# Count missing last-sale dates.
missing_last_sale_dates = (
    sku_master["last_sale_date"]
    .isna()
    .sum()
)

assert missing_last_sale_dates == 0, (
    f"Missing last-sale dates detected: "
    f"{missing_last_sale_dates}."
)

# Calculate sales recency as of the final
# observed historical sales date.
sku_master["days_since_last_sale"] = (
    observation_end_date
    - sku_master["last_sale_date"]
).dt.days.astype("Int32")

# Count invalid negative recency values.
negative_sales_recency = (
    sku_master["days_since_last_sale"] < 0
).sum()

assert negative_sales_recency == 0, (
    f"Negative sales-recency values detected: "
    f"{negative_sales_recency}."
)

# Define the classification conditions.
status_conditions = [
    (
        sku_master["days_since_last_sale"]
        <= ACTIVE_MAX_DAYS
    ).to_numpy(),
    (
        sku_master["days_since_last_sale"]
        <= SLOW_MOVING_MAX_DAYS
    ).to_numpy()
]

# Classify every SKU using the documented rules.
sku_master["sku_status"] = pd.Series(
    np.select(
        status_conditions,
        [
            "Active",
            "Slow Moving"
        ],
        default="Inactive"
    ),
    index=sku_master.index,
    dtype="string"
)

print("✅ SKU status classified successfully.")
print(f"Status reference date : {observation_end_date.date()}")
print(f"Active threshold      : 0–{ACTIVE_MAX_DAYS} days")
print(
    f"Slow Moving threshold : "
    f"{ACTIVE_MAX_DAYS + 1}–{SLOW_MOVING_MAX_DAYS} days"
)
print(
    f"Inactive threshold    : "
    f"More than {SLOW_MOVING_MAX_DAYS} days"
)
print(f"SKUs classified       : {len(sku_master):,}")
````

### 🛠️ 11.4.4 Engineer Inventory Parameters

SKU-level inventory parameters will be estimated from historical average daily demand and the validated simulation assumptions.

This step will calculate reorder points, target stock levels, and fixed replenishment quantities for each simulated SKU.

The validated parameters will guide the sequential inventory simulation.

````python
# ============================================================
# Engineer Demand and Replenishment Parameters
# ============================================================

# Calculate historical average daily demand by SKU.
demand_parameters = (
    inventory_working
    .groupby(
        "sku_id",
        as_index=False
    )
    .agg(
        average_daily_demand=(
            "units_sold",
            "mean"
        )
    )
    .sort_values("sku_id")
    .reset_index(drop=True)
)

# Assign the fixed replenishment lead time.
demand_parameters["lead_time_days"] = (
    LEAD_TIME_DAYS
)

# Reorder point:
# lead-time demand plus safety-stock demand.
demand_parameters["reorder_point"] = (
    np.ceil(
        demand_parameters["average_daily_demand"]
        * (
            LEAD_TIME_DAYS
            + SAFETY_STOCK_DAYS
        )
    )
    .astype("Int32")
    .clip(lower=1)
)

# Target stock:
# demand coverage, lead time and safety stock.
demand_parameters["target_stock"] = (
    np.ceil(
        demand_parameters["average_daily_demand"]
        * TARGET_COVERAGE_DAYS
    )
    .astype("Int32")
    .clip(lower=1)
)

# Fixed replenishment quantity using the
# documented bulk-order multiplier.
demand_parameters["reorder_quantity"] = (
    np.ceil(
        demand_parameters["target_stock"]
        * REORDER_QUANTITY_MULTIPLIER
    )
    .astype("Int32")
    .clip(lower=1)
)

print("✅ Demand and replenishment parameters engineered successfully.")
print(f"Total SKUs : {len(demand_parameters):,}")
````

## 🔄 11.4.5 Simulate Daily Inventory Movements

Daily inventory movements will be simulated chronologically for each SKU using historical demand and the engineered replenishment parameters.

Stock receipts will be processed before demand, while new orders will be placed after evaluating the closing inventory position. Only one replenishment order may remain outstanding for a SKU at a time.

The completed simulation will provide one inventory-movement record for every SKU-date combination.

````python
# ============================================================
# Simulate Daily Inventory Movements
# ============================================================

# Add SKU-level inventory parameters.
inventory_simulation_base = (
    inventory_working
    .merge(
        demand_parameters[
            [
                "sku_id",
                "lead_time_days",
                "reorder_point",
                "target_stock",
                "reorder_quantity"
            ]
        ],
        on="sku_id",
        how="left",
        validate="many_to_one"
    )
    .sort_values(
        ["sku_id", "date"]
    )
    .reset_index(drop=True)
)

# Confirm that all parameters were matched.
missing_simulation_parameters = (
    inventory_simulation_base[
        [
            "lead_time_days",
            "reorder_point",
            "target_stock",
            "reorder_quantity"
        ]
    ]
    .isna()
    .sum()
    .sum()
)

assert missing_simulation_parameters == 0, (
    f"Missing inventory parameters detected: "
    f"{missing_simulation_parameters}."
)

simulation_results = []

# Simulate each SKU independently.
for sku_id, sku_data in inventory_simulation_base.groupby(
    "sku_id",
    sort=False
):
    sku_data = (
        sku_data
        .sort_values("date")
        .reset_index(drop=True)
        .copy()
    )

    lead_time_days = int(
        sku_data["lead_time_days"].iloc[0]
    )

    reorder_point = int(
        sku_data["reorder_point"].iloc[0]
    )

    target_stock = int(
        sku_data["target_stock"].iloc[0]
    )

    reorder_quantity = int(
        sku_data["reorder_quantity"].iloc[0]
    )

    # Initialize inventory at the target-stock level.
    current_stock = target_stock

    # arrival_position: ordered quantity
    pending_orders = {}

    opening_stock_values = []
    fulfilled_units_values = []
    lost_sales_values = []
    received_stock_values = []
    closing_stock_values = []
    order_placed_values = []
    on_order_values = []

    for day_position, daily_demand in enumerate(
        sku_data["units_sold"].to_numpy()
    ):
        daily_demand = int(daily_demand)

        # Previous closing stock becomes today's opening stock.
        opening_stock = current_stock

        # Receive orders scheduled for the current day.
        received_stock = pending_orders.pop(
            day_position,
            0
        )

        available_stock = (
            opening_stock
            + received_stock
        )

        # Fulfil demand using available stock.
        fulfilled_units = min(
            daily_demand,
            available_stock
        )

        lost_sales_units = (
            daily_demand
            - fulfilled_units
        )

        closing_stock = (
            available_stock
            - fulfilled_units
        )

        # Inventory already ordered but not yet received.
        outstanding_order_units = sum(
            pending_orders.values()
        )

        inventory_position = (
            closing_stock
            + outstanding_order_units
        )

        order_placed_units = 0

        # Place an order only when no order is outstanding.
        if (
            inventory_position <= reorder_point
            and outstanding_order_units == 0
        ):
            arrival_position = (
                day_position
                + lead_time_days
            )

            order_placed_units = reorder_quantity

            pending_orders[arrival_position] = (
                order_placed_units
            )

        outstanding_order_units = sum(
            pending_orders.values()
        )

        opening_stock_values.append(opening_stock)
        fulfilled_units_values.append(fulfilled_units)
        lost_sales_values.append(lost_sales_units)
        received_stock_values.append(received_stock)
        closing_stock_values.append(closing_stock)
        order_placed_values.append(order_placed_units)
        on_order_values.append(outstanding_order_units)

        current_stock = closing_stock

    sku_data["opening_stock"] = opening_stock_values
    sku_data["fulfilled_units"] = fulfilled_units_values
    sku_data["lost_sales_units"] = lost_sales_values
    sku_data["received_stock"] = received_stock_values
    sku_data["closing_stock"] = closing_stock_values
    sku_data["order_placed_units"] = order_placed_values
    sku_data["on_order_units"] = on_order_values

    simulation_results.append(sku_data)

# Combine all SKU simulations.
inventory_movements = (
    pd.concat(
        simulation_results,
        ignore_index=True
    )
    .sort_values(
        ["sku_id", "date"]
    )
    .reset_index(drop=True)
)

print("✅ Daily inventory movements simulated successfully.")
print(f"Dataset shape  : {inventory_movements.shape}")
print(
    f"Unique SKUs   : "
    f"{inventory_movements['sku_id'].nunique():,}"
)
print(
    f"Total demand  : "
    f"{inventory_movements['units_sold'].sum():,}"
)
print(
    f"Units fulfilled: "
    f"{inventory_movements['fulfilled_units'].sum():,}"
)
print(
    f"Lost sales    : "
    f"{inventory_movements['lost_sales_units'].sum():,}"
)
````

````python
# ============================================================
# Validate Daily Inventory Movements
# ============================================================

expected_movement_columns = [
    "date",
    "sku_id",
    "units_sold",
    "unit_price",
    "estimated_unit_cost",
    "lead_time_days",
    "reorder_point",
    "target_stock",
    "reorder_quantity",
    "opening_stock",
    "fulfilled_units",
    "lost_sales_units",
    "received_stock",
    "closing_stock",
    "order_placed_units",
    "on_order_units"
]

expected_movement_rows = len(
    inventory_simulation_base
)

actual_movement_rows = len(
    inventory_movements
)

unique_movement_skus = (
    inventory_movements["sku_id"].nunique()
)

missing_movement_values = (
    inventory_movements.isna().sum().sum()
)

duplicate_movement_records = (
    inventory_movements[
        ["sku_id", "date"]
    ]
    .duplicated()
    .sum()
)

movement_records_sorted = (
    pd.MultiIndex.from_frame(
        inventory_movements[
            ["sku_id", "date"]
        ]
    )
    .is_monotonic_increasing
)

# Inventory and demand balances.
invalid_inventory_balances = (
    inventory_movements["closing_stock"]
    != (
        inventory_movements["opening_stock"]
        + inventory_movements["received_stock"]
        - inventory_movements["fulfilled_units"]
    )
).sum()

invalid_demand_balances = (
    inventory_movements["units_sold"]
    != (
        inventory_movements["fulfilled_units"]
        + inventory_movements["lost_sales_units"]
    )
).sum()

# Opening-stock continuity.
previous_closing_stock = (
    inventory_movements
    .groupby("sku_id")["closing_stock"]
    .shift(1)
)

non_initial_records = previous_closing_stock.notna()

invalid_opening_continuity = (
    inventory_movements.loc[
        non_initial_records,
        "opening_stock"
    ].to_numpy()
    != previous_closing_stock.loc[
        non_initial_records
    ].to_numpy()
).sum()

# Initial stock must equal target stock.
initial_inventory_records = (
    inventory_movements
    .groupby("sku_id")
    .head(1)
)

invalid_initial_stock = (
    initial_inventory_records["opening_stock"]
    != initial_inventory_records["target_stock"]
).sum()

# Validate replenishment receipt timing.
expected_received_stock = (
    inventory_movements
    .groupby("sku_id")["order_placed_units"]
    .shift(
        LEAD_TIME_DAYS,
        fill_value=0
    )
)

invalid_receipt_timing = (
    inventory_movements["received_stock"]
    != expected_received_stock
).sum()

# Validate outstanding-order movement.
previous_on_order = (
    inventory_movements
    .groupby("sku_id")["on_order_units"]
    .shift(
        1,
        fill_value=0
    )
)

expected_on_order = (
    previous_on_order
    - inventory_movements["received_stock"]
    + inventory_movements["order_placed_units"]
)

invalid_on_order_balances = (
    inventory_movements["on_order_units"]
    != expected_on_order
).sum()

# Validate order-placement policy.
pre_order_outstanding = (
    previous_on_order
    - inventory_movements["received_stock"]
)

pre_order_inventory_position = (
    inventory_movements["closing_stock"]
    + pre_order_outstanding
)

expected_order_placed = np.where(
    (
        pre_order_inventory_position
        <= inventory_movements["reorder_point"]
    )
    &
    (
        pre_order_outstanding == 0
    ),
    inventory_movements["reorder_quantity"],
    0
)

invalid_order_placements = (
    inventory_movements["order_placed_units"]
    != expected_order_placed
).sum()

negative_inventory_measures = (
    inventory_movements[
        [
            "opening_stock",
            "fulfilled_units",
            "lost_sales_units",
            "received_stock",
            "closing_stock",
            "order_placed_units",
            "on_order_units"
        ]
    ] < 0
).sum().sum()

fulfilled_above_demand = (
    inventory_movements["fulfilled_units"]
    > inventory_movements["units_sold"]
).sum()

# Apply validation rules.
assert inventory_movements.columns.tolist() == (
    expected_movement_columns
)

assert actual_movement_rows == expected_movement_rows
assert unique_movement_skus == INVENTORY_SAMPLE_SIZE
assert missing_movement_values == 0
assert duplicate_movement_records == 0
assert movement_records_sorted
assert invalid_inventory_balances == 0
assert invalid_demand_balances == 0
assert invalid_opening_continuity == 0
assert invalid_initial_stock == 0
assert invalid_receipt_timing == 0
assert invalid_on_order_balances == 0
assert invalid_order_placements == 0
assert negative_inventory_measures == 0
assert fulfilled_above_demand == 0

print("✅ Daily inventory movements validated successfully.")
print(f"Expected records          : {expected_movement_rows:,}")
print(f"Actual records            : {actual_movement_rows:,}")
print(f"Unique SKUs               : {unique_movement_skus:,}")
print(f"Missing values            : {missing_movement_values:,}")
print(f"Duplicate SKU-dates       : {duplicate_movement_records:,}")
print(f"Inventory balance errors  : {invalid_inventory_balances:,}")
print(f"Demand balance errors     : {invalid_demand_balances:,}")
print(f"Continuity errors         : {invalid_opening_continuity:,}")
print(f"Receipt timing errors     : {invalid_receipt_timing:,}")
print(f"Order placement errors    : {invalid_order_placements:,}")
````

## 🛠️ 11.4.6 Engineer Inventory Metrics

The simulated inventory movements will be enriched with inventory-status, estimated inventory-value, and days-of-supply metrics.

These fields will support inventory-health monitoring and downstream reporting.

````python
# ============================================================
# Engineer Inventory Status and Business Metrics
# ============================================================

OVERSTOCK_THRESHOLD_MULTIPLIER = 1.50

# Add historical average daily demand.
inventory_final = (
    inventory_movements
    .merge(
        demand_parameters[
            [
                "sku_id",
                "average_daily_demand"
            ]
        ],
        on="sku_id",
        how="left",
        validate="many_to_one"
    )
    .copy()
)

# Confirm valid demand values before calculating supply coverage.
missing_average_demand = (
    inventory_final["average_daily_demand"]
    .isna()
    .sum()
)

non_positive_average_demand = (
    inventory_final["average_daily_demand"] <= 0
).sum()

assert missing_average_demand == 0
assert non_positive_average_demand == 0

# Calculate estimated closing-inventory value.
inventory_final["estimated_inventory_value"] = (
    inventory_final["closing_stock"]
    * inventory_final["estimated_unit_cost"]
).round(2)

# Calculate days of supply.
inventory_final["days_of_supply"] = (
    inventory_final["closing_stock"]
    / inventory_final["average_daily_demand"]
).round(1)

# Classify inventory status.
inventory_final["inventory_status"] = pd.Series(
    np.select(
        [
            (
                inventory_final["closing_stock"] == 0
            ).to_numpy(),

            (
                inventory_final["closing_stock"]
                <= inventory_final["reorder_point"]
            ).to_numpy(),

            (
                inventory_final["closing_stock"]
                >= (
                    inventory_final["target_stock"]
                    * OVERSTOCK_THRESHOLD_MULTIPLIER
                )
            ).to_numpy()
        ],
        [
            "Stockout",
            "Low Stock",
            "Overstock"
        ],
        default="Healthy"
    ),
    index=inventory_final.index,
    dtype="string"
)

print("✅ Inventory status and business metrics engineered successfully.")
print(f"Dataset shape      : {inventory_final.shape}")
print(
    f"Inventory statuses : "
    f"{inventory_final['inventory_status'].nunique()}"
)
````

## ✅ 12.3 Validate Cross-Dataset Consistency

This step verifies Calendar coverage, SKU Master coverage, shared SKU-date keys, and demand consistency between Sales Daily and Inventory Snapshot.

Successful validation confirms that the engineered datasets form a consistent analytical data model.

````python
# ============================================================
# Validate Cross-Dataset Consistency
# ============================================================

# ------------------------------------------------------------
# Prepare Normalized References
# ------------------------------------------------------------

calendar_date_reference = (
    pd.to_datetime(calendar_final["date"])
    .dt.normalize()
)

sku_master_reference = (
    sku_master_final["sku_id"]
    .astype("string")
)

sales_reference = (
    sales_daily_final[
        [
            "sku_id",
            "date",
            "units_sold"
        ]
    ]
    .copy()
)

sales_reference["sku_id"] = (
    sales_reference["sku_id"]
    .astype("string")
)

sales_reference["date"] = (
    pd.to_datetime(sales_reference["date"])
    .dt.normalize()
)

inventory_reference = (
    inventory_snapshots_final[
        [
            "sku_id",
            "date",
            "units_sold"
        ]
    ]
    .copy()
)

inventory_reference["sku_id"] = (
    inventory_reference["sku_id"]
    .astype("string")
)

inventory_reference["date"] = (
    pd.to_datetime(inventory_reference["date"])
    .dt.normalize()
)


# ------------------------------------------------------------
# Validate Key Uniqueness
# ------------------------------------------------------------

calendar_key_unique = (
    calendar_date_reference
    .duplicated()
    .sum()
    == 0
)

sku_master_key_unique = (
    sku_master_reference
    .duplicated()
    .sum()
    == 0
)

sales_key_unique = (
    sales_reference[
        [
            "sku_id",
            "date"
        ]
    ]
    .duplicated()
    .sum()
    == 0
)

inventory_key_unique = (
    inventory_reference[
        [
            "sku_id",
            "date"
        ]
    ]
    .duplicated()
    .sum()
    == 0
)


# ------------------------------------------------------------
# Validate Date Coverage
# ------------------------------------------------------------

calendar_dates = set(
    calendar_date_reference
)

sales_dates = set(
    sales_reference["date"]
)

inventory_dates = set(
    inventory_reference["date"]
)

sales_calendar_coverage = (
    sales_dates.issubset(calendar_dates)
)

inventory_calendar_coverage = (
    inventory_dates.issubset(calendar_dates)
)

sales_inventory_date_alignment = (
    sales_dates == inventory_dates
)


# ------------------------------------------------------------
# Validate SKU Coverage
# ------------------------------------------------------------

sku_master_skus = set(
    sku_master_reference
)

sales_skus = set(
    sales_reference["sku_id"]
)

inventory_skus = set(
    inventory_reference["sku_id"]
)

sales_inventory_sku_alignment = (
    sales_skus == inventory_skus
)

sales_sku_master_coverage = (
    sales_skus.issubset(
        sku_master_skus
    )
)

inventory_sku_master_coverage = (
    inventory_skus.issubset(
        sku_master_skus
    )
)


# ------------------------------------------------------------
# Validate Shared SKU-Date Business Keys
# ------------------------------------------------------------

sales_keys = (
    sales_reference[
        [
            "sku_id",
            "date"
        ]
    ]
    .drop_duplicates()
)

inventory_keys = (
    inventory_reference[
        [
            "sku_id",
            "date"
        ]
    ]
    .drop_duplicates()
)

key_comparison = (
    sales_keys
    .merge(
        inventory_keys,
        on=[
            "sku_id",
            "date"
        ],
        how="outer",
        indicator=True
    )
)

sales_only_keys = (
    key_comparison["_merge"]
    .eq("left_only")
    .sum()
)

inventory_only_keys = (
    key_comparison["_merge"]
    .eq("right_only")
    .sum()
)

business_key_alignment = (
    sales_only_keys == 0
    and inventory_only_keys == 0
    and len(sales_keys) == len(inventory_keys)
)


# ------------------------------------------------------------
# Validate Historical Demand Preservation
# ------------------------------------------------------------

demand_comparison = (
    sales_reference
    .merge(
        inventory_reference,
        on=[
            "sku_id",
            "date"
        ],
        how="inner",
        suffixes=(
            "_sales",
            "_inventory"
        )
    )
)

demand_value_mismatches = (
    demand_comparison["units_sold_sales"]
    != demand_comparison["units_sold_inventory"]
).sum()

units_sold_alignment = (
    business_key_alignment
    and len(demand_comparison) == len(sales_reference)
    and demand_value_mismatches == 0
)


# ------------------------------------------------------------
# Create Validation Summary
# ------------------------------------------------------------

cross_dataset_checks = {
    "Calendar date key is unique":
        calendar_key_unique,

    "SKU Master key is unique":
        sku_master_key_unique,

    "Sales SKU-date key is unique":
        sales_key_unique,

    "Inventory SKU-date key is unique":
        inventory_key_unique,

    "Sales dates covered by Calendar":
        sales_calendar_coverage,

    "Inventory dates covered by Calendar":
        inventory_calendar_coverage,

    "Sales and Inventory dates align":
        sales_inventory_date_alignment,

    "Sales and Inventory SKUs align":
        sales_inventory_sku_alignment,

    "Sales SKUs covered by SKU Master":
        sales_sku_master_coverage,

    "Inventory SKUs covered by SKU Master":
        inventory_sku_master_coverage,

    "Sales and Inventory keys align":
        business_key_alignment,

    "Sales demand preserved in Inventory":
        units_sold_alignment
}

cross_dataset_summary = pd.DataFrame({
    "Validation Check": list(
        cross_dataset_checks.keys()
    ),
    "Status": [
        "Passed" if result else "Failed"
        for result in cross_dataset_checks.values()
    ]
})

failed_cross_dataset_checks = [
    check_name
    for check_name, result
    in cross_dataset_checks.items()
    if not result
]


# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("Cross-Dataset Consistency Diagnostics")
print("-" * 48)
print(f"Sales-only SKU-date keys     : {sales_only_keys:,}")
print(f"Inventory-only SKU-date keys : {inventory_only_keys:,}")
print(f"Demand value mismatches      : {demand_value_mismatches:,}")

print("\nCross-Dataset Validation Summary\n")

display(
    cross_dataset_summary
)

assert len(failed_cross_dataset_checks) == 0, (
    f"Cross-dataset validation failed: "
    f"{failed_cross_dataset_checks}."
)

print("✅ Cross-dataset consistency validated successfully.")
````

#### ✅ Validation Result

Calendar coverage, SKU Master coverage, shared SKU-date keys, and historical demand values were fully aligned.

The engineered datasets form a consistent analytical data model.
